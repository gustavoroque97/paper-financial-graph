```python
import pandas as pd 
import numpy as np
from datetime import date

returns = pd.read_parquet("../../data/01_raw/synthetic_returns.parquet")
returns = returns.loc[:, ~returns.isna().all()]

# Dividing by years
for year in range(2019,2025):
    returns_year = returns.loc[f"{year-1}-01-01":f"{year}-12-31"]
    returns_year.to_parquet(f"../../data/02_clean/synthetic_returns_{year-1}_{year}.parquet")
```
